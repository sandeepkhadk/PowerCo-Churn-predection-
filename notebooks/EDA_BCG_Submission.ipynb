## Data visualization

If you're working in Python, two of the most popular packages for visualization are `matplotlib` and `seaborn`. We highly recommend you use these, or at least be familiar with them because they are ubiquitous!

Below are some functions that you can use to get started with visualizations. 

```python
def plot_stacked_bars(dataframe, title_, size_=(18, 10), rot_=0, legend_="upper right"):
    """
    Plot stacked bars with annotations
    """
    ax = dataframe.plot(
        kind="bar",
        stacked=True,
        figsize=size_,
        rot=rot_,
        title=title_
    )

    # Annotate bars
    annotate_stacked_bars(ax, textsize=14)
    # Rename legend
    plt.legend(["Retention", "Churn"], loc=legend_)
    # Labels
    plt.ylabel("Company base (%)")
    plt.show()

def annotate_stacked_bars(ax, pad=0.99, colour="white", textsize=13):
    """
    Add value annotations to the bars
    """

    # Iterate over the plotted rectanges/bars
    for p in ax.patches:
        
        # Calculate annotation
        value = str(round(p.get_height(),1))
        # If value is 0 do not annotate
        if value == '0.0':
            continue
        ax.annotate(
            value,
            ((p.get_x()+ p.get_width()/2)*pad-0.05, (p.get_y()+p.get_height()/2)*pad),
            color=colour,
            size=textsize
        )

def plot_distribution(dataframe, column, ax, bins_=50):
    """
    Plot variable distirbution in a stacked histogram of churned or retained company
    """
    # Create a temporal dataframe with the data to be plot
    temp = pd.DataFrame({"Retention": dataframe[dataframe["churn"]==0][column],
    "Churn":dataframe[dataframe["churn"]==1][column]})
    # Plot the histogram
    temp[["Retention","Churn"]].plot(kind='hist', bins=bins_, ax=ax, stacked=True)
    # X-axis label
    ax.set_xlabel(column)
    # Change the x-axis to plain style
    ax.ticklabel_format(style='plain', axis='x')
```

Thhe first function `plot_stacked_bars` is used to plot a stacked bar chart. An example of how you could use this is shown below:

```python
churn = client_df[['id', 'churn']]
churn.columns = ['Companies', 'churn']
churn_total = churn.groupby(churn['churn']).count()
churn_percentage = churn_total / churn_total.sum() * 100
plot_stacked_bars(churn_percentage.transpose(), "Churning status", (5, 5), legend_="lower right")
```

The second function `annotate_bars` is used by the first function, but the third function `plot_distribution` helps you to plot the distribution of a numeric column. An example of how it can be used is given below:

```python
consumption = client_df[['id', 'cons_12m', 'cons_gas_12m', 'cons_last_month', 'imp_cons', 'has_gas', 'churn']]

fig, axs = plt.subplots(nrows=1, figsize=(18, 5))

plot_distribution(consumption, 'cons_12m', axs)
```

## 4. Data types, unique values and missing values

The following table summarizes the data types, number of unique values, and pandas-level missing values for each customer variable. This helps us understand the structure and quality of the dataset before further analysis.

```python
client_data_info = pd.DataFrame({
    "dtype": client_df.dtypes.astype(str),
    "unique_values": client_df.nunique(),
    "missing_values": client_df.isnull().sum()
}).sort_values("unique_values")

display(client_data_info)
```

## 5. Categorical variables

Categorical variables are examined using frequency counts. High-cardinality fields such as IDs and dates are not plotted as bar charts because they would not provide a useful visualization.

```python
categorical_cols = client_df.select_dtypes(include="object").columns.tolist()

for col in categorical_cols:
    if client_df[col].nunique() <= 15:
        plt.figure(figsize=(8, 4))
        order = client_df[col].value_counts().index
        sns.countplot(data=client_df, x=col, order=order)
        plt.title(f"Distribution of {col}")
        plt.xticks(rotation=45, ha="right")
        plt.tight_layout()
        plt.show()
```

## 6. Churn distribution

Churn is the main outcome variable in the customer dataset. We examine its frequency and percentage distribution.

```python
churn_summary = client_df["churn"].value_counts().sort_index().to_frame("count")
churn_summary["percentage"] = (churn_summary["count"] / len(client_df) * 100).round(2)
display(churn_summary)
```

```python
plt.figure(figsize=(6, 4))
ax = sns.countplot(data=client_df, x="churn")
plt.title("Customer Churn Distribution")
plt.xlabel("Churn")
plt.ylabel("Number of Customers")
for container in ax.containers:
    ax.bar_label(container)
plt.tight_layout()
plt.show()
```

## 7. Numerical variables by churn status

The following plots compare selected customer characteristics for churned and non-churned customers. These are exploratory comparisons and do not imply causation.

```python
selected_features = [
    "cons_12m", "cons_gas_12m", "forecast_cons_12m",
    "net_margin", "num_years_antig", "nb_prod_act"
]

for col in selected_features:
    if col in client_df.columns:
        plt.figure(figsize=(7, 4))
        sns.boxplot(data=client_df, x="churn", y=col)
        plt.title(f"{col} by Churn Status")
        plt.xlabel("Churn")
        plt.ylabel(col)
        plt.tight_layout()
        plt.show()
```

## 8. Historical pricing data

The pricing dataset is explored using descriptive statistics, distributions, and average price trends over time.

```python
price_info = pd.DataFrame({
    "dtype": price_df.dtypes.astype(str),
    "unique_values": price_df.nunique(),
    "missing_values": price_df.isnull().sum()
})
display(price_info)

display(price_df.describe())
```

```python
numerical_price_cols = price_df.select_dtypes(include="number").columns.tolist()
price_df[numerical_price_cols].hist(figsize=(16, 12), bins=30, edgecolor="black")
plt.suptitle("Distributions of Historical Pricing Variables", y=1.02, fontsize=16)
plt.tight_layout()
plt.show()
```

```python
price_df["price_date"] = pd.to_datetime(price_df["price_date"], errors="coerce")
price_by_date = (price_df.groupby("price_date").mean(numeric_only=True).reset_index().sort_values("price_date"))

price_columns = ["price_p1_var", "price_p1_fix", "price_p2_var", "price_p2_fix", "price_p3_var", "price_p3_fix"]
for col in price_columns:
    if col in price_by_date.columns:
        plt.figure(figsize=(10, 4))
        plt.plot(price_by_date["price_date"], price_by_date[col])
        plt.title(f"{col} Over Time")
        plt.xlabel("Date")
        plt.ylabel(col)
        plt.tight_layout()
        plt.show()
```
